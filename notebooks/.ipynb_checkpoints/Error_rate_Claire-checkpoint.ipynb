```python
def contains_step_error(lst, sub):
    """
    function takes as input a list and looks for a sublist inside
    input: list and sub list
    output boolean
    """
    n = len(sub)
    return any(lst[i:i+n] == sub for i in range(len(lst)-n+1))
```

```python
steps = ['start', 'step_1', 'step_2', 'step_3', 'confirm']

def get_completion_rates(df):
    total = df['client_id'].nunique()
    rates = {}
    for step in steps:
        clients_reached = df[df['process_step'] == step]['client_id'].nunique()
        rates[step] = clients_reached #/ total
    return rates
```

```python
def get_error_rate(df):
    patterns = {
        's_1': ['start', 'step_1', 'start'],
        '1_2': ['step_1', 'step_2', 'step_1'],
        '2_3': ['step_2', 'step_3', 'step_2'],
        '3_c': ['step_3', 'confirm', 'step_3']
    }
    results = {key: set() for key in patterns}
    err_ids = set()
    
    for client_id, group in df.groupby('client_id'):
        step_list = group['process_step'].tolist()
        for key, pattern in patterns.items():
            if contains_step_error(step_list, pattern):
                results[key].add(client_id)
                err_ids.add(client_id)
    
    # total = df['client_id'].nunique()
    # err_pc = len(err_ids)/total*100
    # error_rates = {key: len(ids) / total for key, ids in results.items()}
    error_rates = {key: len(ids) for key, ids in results.items()}
    error_rates['err_counts']=len(err_ids)   #err_pc
    results['err_counts']=err_ids
    return error_rates, results

    #working with counts as the denominator is different across steps
```

```python
from scipy.stats import chi2_contingency
import numpy as np

# within control
ctrl_counts = list(finished_cont.values())
ctrl_dropped = [ctrl_counts[i-1] - ctrl_counts[i] for i in range(1, len(ctrl_counts))]
ctrl_continued = ctrl_counts[1:]

stat, pval, dof, expected = chi2_contingency([ctrl_continued, ctrl_dropped])
print(f"Control within-group: chi2={stat:.3f}, p={pval:.4f}")

# within test
test_counts = list(finished_test.values())
test_dropped = [test_counts[i-1] - test_counts[i] for i in range(1, len(test_counts))]
test_continued = test_counts[1:]

stat, pval, dof, expected = chi2_contingency([test_continued, test_dropped])
print(f"Test within-group: chi2={stat:.3f}, p={pval:.4f}")
```

```python
df_cd_c = df_cd[df_cd.client_id.isin(df_cont.client_id)]
df_cd_t = df_cd[df_cd.client_id.isin(df_test.client_id)]
```

```python
df_wcontrol['date_time'] = pd.to_datetime(df_wcontrol['date_time'])
df_wtest['date_time'] = pd.to_datetime(df_wtest['date_time'])
```

```python
df_cd_c['group'] = 'control'
df_cd_t['group'] = 'test'

def get_completion_time(group):
    starts = group[group['process_step'] == 'start']['date_time']
    confirms = group[group['process_step'] == 'confirm']['date_time']
    if starts.empty or confirms.empty:
        return None
    return (confirms.max() - starts.max()) #.total_seconds()

def add_error_cols(df_cd, df_w, results):
    error_ids = results['err_counts']
    pattern_keys = ['s_1', '1_2', '2_3', '3_c']
    
    completion_times = df_w.groupby('client_id').apply(get_completion_time).reset_index()
    completion_times.columns = ['client_id', 'completion_time']
    
    df_cd = df_cd.merge(completion_times, on='client_id', how='left')
    df_cd['made_error'] = df_cd['client_id'].isin(error_ids)
    df_cd['n_errors'] = df_cd['client_id'].apply(
        lambda x: sum(x in results[key] for key in pattern_keys)
    )
    return df_cd

df_cd_c = add_error_cols(df_cd_c, df_wcontrol, res_cont)
df_cd_t = add_error_cols(df_cd_t, df_wtest, res_test)
df_cd = pd.concat([df_cd_c, df_cd_t], ignore_index=True)
```

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

def ztest_by_age(df, group_col="group", outcome="made_error"):
    results = []

    for age_bin_q, sub in df.groupby("age_bin_q"):
        test = sub[sub[group_col] == "test"]
        control = sub[sub[group_col] == "control"]

        if len(test) == 0 or len(control) == 0:
            continue

        count = np.array([
            test[outcome].sum(),
            control[outcome].sum()
        ])

        nobs = np.array([
            len(test),
            len(control)
        ])

        stat, pval = proportions_ztest(count, nobs)

        results.append({
            "age_bin": age_bin_q,
            "test_rate": test[outcome].mean(),
            "control_rate": control[outcome].mean(),
            "z_stat": stat,
            "p_value": pval
        })

    return pd.DataFrame(results)
```

```python
def mean_diff_ztest(df, outcome="n_errors"):
    from scipy import stats
    results = []

    for age_bin_q, sub in df.groupby("age_bin_q"):
        test = sub[sub["group"] == "test"][outcome].dropna()
        control = sub[sub["group"] == "control"][outcome].dropna()

        if len(test) < 2 or len(control) < 2:
            continue

        mean1, mean2 = test.mean(), control.mean()
        var1, var2 = test.var(ddof=1), control.var(ddof=1)

        n1, n2 = len(test), len(control)

        se = np.sqrt(var1/n1 + var2/n2)
        z = (mean1 - mean2) / se if se > 0 else np.nan

        p = 2 * (1 - stats.norm.cdf(abs(z))) if se > 0 else np.nan

        results.append({
            "age_bin": age_bin_q,
            "test_mean": mean1,
            "control_mean": mean2,
            "z_stat": z,
            "p_value": p
        })

    return pd.DataFrame(results)
```

```python
import numpy as np
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest

results = []

for age_bin, sub in df_cd.groupby("age_bin_q"):
    test = sub[sub["group"] == "test"]
    control = sub[sub["group"] == "control"]

    if len(test) == 0 or len(control) == 0:
        continue

    count = np.array([
        test["completed"].sum(),
        control["completed"].sum()
    ])

    nobs = np.array([
        len(test),
        len(control)
    ])

    z_stat, p_value = proportions_ztest(count, nobs)

    results.append({
        "age_bin": age_bin,
        "test_rate": test["completed"].mean(),
        "control_rate": control["completed"].mean(),
        "abs_diff": test["completed"].mean() - control["completed"].mean(),
        "z_stat": z_stat,
        "p_value": p_value
    })

res_completed_q = pd.DataFrame(results)
res_completed_q
```
