## removing same steps duplicated

```python
df_sorted = df_final_web_data.sort_values(by=["visit_id", "date_time"], ascending=True)

same_step_as_prev = (df_sorted["process_step"] == df_sorted["process_step"].shift(1)) & (df_sorted["visit_id"] == df_sorted["visit_id"].shift(1))
same_step_as_next = (df_sorted["process_step"] == df_sorted["process_step"].shift(-1)) & (df_sorted["visit_id"] == df_sorted["visit_id"].shift(-1))

drop_confirm = (df_sorted["process_step"] == "confirm") & same_step_as_prev

drop_start = (df_sorted["process_step"] == "start") & same_step_as_next

drop_steps = df_sorted["process_step"].isin(["step_1", "step_2", "step_3", "step_4"]) & same_step_as_prev


drop_start = (df_sorted["process_step"] == "start") & same_step_as_next
drop_steps = df_sorted["process_step"].isin(["step_1", "step_2", "step_3", "step_4"]) & same_step_as_prev

rows_to_delete = drop_confirm | drop_start | drop_steps

df_clean = df_sorted[~rows_to_delete]

df_final = df_clean.sort_values(by=["visit_id", "date_time"], ascending=[True, False]).reset_index(drop=True)

display(df_final.shape)
df_final.info()
df_final.head(40)
display(df_final)
```

```python
df_final["date_time"] = pd.to_datetime(df_final["date_time"])
df_final.dtypes
```

## Only confirmed

```python
df_confirm = df_final[df_final["process_step"] == "confirm"]

display(df_confirm.info)

df_confirm.head(20)
```

## pivot to see nan

```python
df_pivot = df_final.pivot_table(
    index=["visitor_id", "visit_id"],
    columns="process_step",
    values="date_time" ,
    aggfunc="min")

steps = ["start", "step_1", "step_2", "step_3", "confirm"]

df_pivot = df_pivot.reindex(columns=steps)

df_final_pivot = df_pivot.reset_index()

display(df_final_pivot)
```

## compleition rate

```python
unique_visits = df_final_pivot["visit_id"].nunique()

total_confirms_pivot = df_final_pivot["confirm"].notnull().sum()

conversion_rate = (total_confirms_pivot / unique_visits) * 100

print(f"Total visits (Pivot): {unique_visits:,}")
print(f"Total confirms (Pivot): {total_confirms_pivot:,}")
print(f"confirmation rate: {conversion_rate:.2f}%")
```
